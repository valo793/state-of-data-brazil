# 01. Análise Exploratória de Metadados — Camada Bronze

Este notebook realiza a inspeção preliminar dos dados brutos das 3 edições da pesquisa **State of Data Brasil** armazenadas na Camada Bronze (`data/bronze`).

```python
import pandas as pd
from pathlib import Path

bronze_dir = Path("../data/bronze")
files = sorted(list(bronze_dir.glob("*.csv")))
print(f"Total de arquivos encontrados: {len(files)}")
for f in files:
    print(f" - {f.name}")
```

## 1. Carregamento e Inspeção de Esquemas

```python
for f in files:
    df = pd.read_csv(f, low_memory=False)
    print(f"Arquivo: {f.name}")
    print(f"  Shape: {df.shape[0]} linhas x {df.shape[1]} colunas\n")
```

## 2. Diagnóstico de Qualidade e Estrutura dos Cabeçalhos

Observação da diferença entre as tuplas de 2023-2024 e os nomes descritivos de 2024+.

```python
df23 = pd.read_csv(files[0], nrows=3)
print("Amostra de colunas 2023-2024:")
print(list(df23.columns[:5]))

df24 = pd.read_csv(files[1], nrows=3)
print("\nAmostra de colunas 2024-2025:")
print(list(df24.columns[:5]))
```
